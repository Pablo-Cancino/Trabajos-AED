### Estructura de la pila enlazada

La pila usa una **lista enlazada** donde:
- El **tope** de la pila es el **primer nodo** de la lista
- Cada nodo tiene `dato` y una referencia al `siguiente` nodo
- Cuando la pila está vacía, el tope es `None`

```
Diagrama: push(10), push(20), push(30)

   tope
    |
    v
  [30|--] → [20|--] → [10|--] → [None]
   ^
   |
tope = primer nodo de la lista
```

````python
# Esta clase Nodo ya está implementada para ti
class _Nodo:
    """Nodo para lista enlazada. ¡Ya implementada, no modificar!"""
    def __init__(self, dato: Any, siguiente: '_Nodo | None' = None) -> None:
        self.dato = dato
        self.siguiente = siguiente
    
    def __repr__(self) -> str:
        return f"Nodo({self.dato})"

print("✅ Clase _Nodo cargada")
````

````python
class MiPilaEnlazada(PilaTDA):
    """
    Pila implementada con lista enlazada simple.
    El tope de la pila es el primer nodo de la lista.
    """
    
    def __init__(self) -> None:
        self.__tope = None
        self.__tam = 0 
    
    def push(self, item: Any) -> None:
        nuevo_nodo = _Nodo(item, siguiente = self.__tope)
        self.__tope = nuevo_nodo  
        self.__tam += 1
    
    def pop(self) -> Any:
        if self.is_empty(): 
            raise IndexError("pop de pila vacía")
        
        dato_retorno = self.__tope.dato
        self.__tope = self.__tope.siguiente  
        self.__tam -= 1
        return dato_retorno
    
    def peek(self) -> Any:
        if self.is_empty():  
            raise IndexError("peek de pila vacía")

        return self.__tope.dato        
    
    def is_empty(self) -> bool:
        return self.__tam == 0
    
    def size(self) -> int:
        return self.__tam  

    def __repr__(self) -> str:
        if self.is_empty():
            return "MiPilaEnlazada(vacía)"
        
        elementos = []
        actual = self.__tope
        while actual is not None:
            elementos.append(str(actual.dato))
            actual = actual.siguiente
            
        cadena_nodos = " → ".join(elementos)
        return f"MiPilaEnlazada(tope → {cadena_nodos})"
    
    def __len__(self) -> int:
        return self.size()

print("✅ Clase MiPilaEnlazada definida correctamente")
````

````python
# ═══ NO MODIFICAR ═══
# Benchmark: PilaArreglo vs MiPilaEnlazada

import time

N = 10_000

def benchmark(clase_pila, nombre: str, n: int) -> float:
    p = clase_pila()
    inicio = time.perf_counter()
    for i in range(n):
        p.push(i)
    for _ in range(n):
        p.pop()
    fin = time.perf_counter()
    ms = (fin - inicio) * 1000
    print(f"  {nombre:25s}: {ms:.3f} ms para {n:,} push + {n:,} pop")
    return ms

print("─" * 55)
print("BENCHMARK: PilaArreglo vs MiPilaEnlazada")
print("─" * 55)
t1 = benchmark(PilaArreglo, "PilaArreglo (referencia)", N)
t2 = benchmark(MiPilaEnlazada, "MiPilaEnlazada (tuya)", N)
ratio = t2 / t1 if t1 > 0 else 0
print(f"\n  Ratio MiPilaEnlazada / PilaArreglo: {ratio:.2f}x")
print("─" * 55)
````
