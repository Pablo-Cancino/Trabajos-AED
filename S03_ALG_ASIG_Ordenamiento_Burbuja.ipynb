## 📖 Parte 1: ¿Cómo funciona el Ordenamiento por Burbuja?

El **ordenamiento por burbuja** (*Bubble Sort*) es uno de los algoritmos de ordenamiento más simples e intuitivos que existen. Su nombre proviene de la analogía con burbujas en el agua: los elementos más grandes "burbujean" hacia el final de la lista, pasaje a pasaje, de la misma manera en que las burbujas de aire ascienden a la superficie.

### Idea general

El algoritmo recorre la lista de elementos repetidamente. En cada recorrido, compara **pares de elementos adyacentes** y los **intercambia** si están en el orden incorrecto (es decir, si el elemento de la izquierda es mayor que el de la derecha). Este proceso se repite hasta que ningún intercambio sea necesario, lo que indica que la lista ya está ordenada.

### Descripción paso a paso

Supongamos que tenemos el siguiente arreglo de 5 elementos:

```
[ 5,  3,  8,  1,  4 ]
```

**Pasada 1:**
- Comparamos el par (5, 3) → 5 > 3, los intercambiamos → `[ 3, 5, 8, 1, 4 ]`
- Comparamos el par (5, 8) → 5 < 8, no se intercambian → `[ 3, 5, 8, 1, 4 ]`
- Comparamos el par (8, 1) → 8 > 1, los intercambiamos → `[ 3, 5, 1, 8, 4 ]`
- Comparamos el par (8, 4) → 8 > 4, los intercambiamos → `[ 3, 5, 1, 4, 8 ]`
- ✅ El elemento más grande (8) ya está en su posición final.

**Pasada 2:**
- Comparamos (3, 5) → no se intercambia
- Comparamos (5, 1) → intercambiamos → `[ 3, 1, 5, 4, 8 ]`
- Comparamos (5, 4) → intercambiamos → `[ 3, 1, 4, 5, 8 ]`
- ✅ El segundo elemento más grande (5) está en su posición final.

**Pasada 3:**
- Comparamos (3, 1) → intercambiamos → `[ 1, 3, 4, 5, 8 ]`
- Comparamos (3, 4) → no se intercambia
- ✅ El tercer elemento más grande (4) está en su posición final.

**Pasada 4:**
- Comparamos (1, 3) → no se intercambia
- ✅ La lista ya está ordenada: `[ 1, 3, 4, 5, 8 ]`

### Observaciones clave

- Después de cada pasada completa, el elemento más grande de la porción no ordenada queda en su posición correcta al final. Por eso, en cada pasada siguiente se necesita comparar un elemento menos.
- Para un arreglo de **n** elementos, el algoritmo realiza a lo sumo **n-1 pasadas**.
- En el **peor caso** (arreglo ordenado en orden inverso), se realizan aproximadamente $\frac{n(n-1)}{2}$ comparaciones, lo que da una complejidad de $\mathcal{O}(n^2)$.
- En el **mejor caso** (arreglo ya ordenado), una versión optimizada del algoritmo puede detectar que no hubo intercambios y terminar en una sola pasada, logrando $\mathcal{O}(n)$.

### Pseudocódigo

```
procedimiento burbuja(A, n):
    para i desde 0 hasta n-2:
        para j desde 0 hasta n-2-i:
            si A[j] > A[j+1]:
                intercambiar A[j] y A[j+1]
```

> 💡 **Nota:** En el pseudocódigo, el índice `i` representa la pasada actual. A medida que `i` aumenta, el bucle interno recorre un rango más corto porque los últimos `i` elementos ya están en su posición correcta.

## 🔧 Parte 2: Generación de Datos y Medición de Tiempos

En esta sección se proporciona el código necesario para:
1. Generar arreglos de números enteros aleatorios de distintos tamaños.
2. Medir el tiempo de ejecución del algoritmo usando la librería `time`.

**Lee atentamente el código antes de pasar a la siguiente sección.**

````python
# ============================================================
# Importación de librerías necesarias
# ============================================================

import random
import time
import math
import matplotlib.pyplot as plt
import numpy as np

print("✅ Librerías importadas correctamente.")
````

````python
# ============================================================
# Función auxiliar: generar un arreglo aleatorio
# ============================================================
def generar_arreglo(n, valor_min=1, valor_max=100000):
    """
    Genera una lista de n números enteros aleatorios.
    
    Parámetros:
    -----------
    n         : tamaño del arreglo
    valor_min : valor mínimo posible (por defecto 1)
    valor_max : valor máximo posible (por defecto 100000)
    
    Retorna:
    --------
    Una lista de n enteros aleatorios.
    """
    return [random.randint(valor_min, valor_max) for _ in range(n)]

# ============================================================
# Función auxiliar: medir tiempo de ejecución
# ============================================================
def medir_tiempo(funcion_ordenamiento, arreglo):
    """
    Mide el tiempo de ejecución de una función de ordenamiento.
    
    Parámetros:
    -----------
    funcion_ordenamiento : función que recibe un arreglo y lo ordena
    arreglo              : lista de elementos a ordenar
    
    Retorna:
    --------
    tiempo_segundos : tiempo de ejecución en segundos (float)
    """
    # Hacemos una copia para no modificar el arreglo original
    copia = arreglo.copy()
    
    # Registramos el tiempo antes de ordenar
    inicio = time.time()
    
    # Ejecutamos el algoritmo
    funcion_ordenamiento(copia)
    
    # Registramos el tiempo después de ordenar
    fin = time.time()
    
    # Calculamos la diferencia
    tiempo_segundos = fin - inicio
    return tiempo_segundos


# ============================================================
# Definición de los tamaños de arreglo a probar
# IMPORTANTE: se usa una secuencia de duplicación (cada tamaño
# es el doble del anterior), requerida por el método de Sedgewick.
# ============================================================
tamanios = [1000, 2000, 4000, 8000, 16000]

print("✅ Funciones auxiliares definidas.")
print(f"   Tamaños a probar: {tamanios}")
````

## ✏️ Parte 3: Implementa tu Algoritmo de Burbuja

**Instrucciones:**

En la celda de código a continuación, implementa la función `burbuja(A)` que ordene la lista `A` **en su lugar** (*in-place*) usando el algoritmo de ordenamiento por burbuja.

- La función **no debe retornar** nada (modifica el arreglo directamente).
- Usa el pseudocódigo de la Parte 1 como guía.
- Asegúrate de que tu función sea correcta antes de medir los tiempos.

> ⚠️ **No uses** `list.sort()` ni ninguna función de ordenamiento de Python. Debes implementar el algoritmo desde cero.

````python
# ============================================================
# Implementado aquí mi algoritmo de ordenamiento burbuja
# ============================================================

def burbuja(A):
    """
    Ordena el arreglo A en orden ascendente usando el algoritmo
    de ordenamiento por burbuja.
    
    Parámetro:
    ----------
    A : lista de elementos comparables
    
    Retorna:
    --------
    None (modifica A in-place)
    """
    n = len(A)

    for i in range(n):

        #una vez el programa detecte falso, (o sea, que no se haga ningun cambio) se hace break:
        intercambio = False

        for j in range(n-1-i):
            if A[j] > A[j+1]:
                placeholder = A[j+1]
                A[j+1] = A[j]
                A[j] = placeholder
                intercambio = True

        if not intercambio:
            break

print("Declaración de bubble correcta.")
````

````python
# ============================================================
# Prueba de corrección: verifica que tu algoritmo funciona
# antes de medir tiempos con arreglos grandes
# ============================================================

arreglo_prueba = [64, 34, 25, 12, 22, 11, 90]
copia_prueba = arreglo_prueba.copy()
burbuja(copia_prueba)

print("Arreglo original: ", arreglo_prueba)
print("Arreglo ordenado: ", copia_prueba)
print("Resultado esperado: [11, 12, 22, 25, 34, 64, 90]")

if copia_prueba == sorted(arreglo_prueba):
    print("\n✅ ¡Tu algoritmo es correcto! Puedes continuar.")
else:
    print("\n❌ Tu algoritmo NO produce el resultado correcto. Revisa la implementación.")
````

## 🔬 Parte 6: Hipótesis de Duplicación de Sedgewick

### ¿Qué es la hipótesis de duplicación?

La estrategia central de Sedgewick (descrita en *Algorithms, 4th Edition*) para determinar empíricamente el orden de crecimiento de un algoritmo se llama **experimento de duplicación** (*doubling experiment*). Consiste en:

1. Ejecutar el algoritmo con tamaños **N, 2N, 4N, 8N, ...** (secuencia de duplicación).
2. Calcular la **razón de duplicación**: $R = \dfrac{T(2N)}{T(N)}$
3. Estimar el exponente de complejidad mediante:

$$b = \log_2\!\left(\frac{T(2N)}{T(N)}\right)$$

### ¿Por qué funciona?

Si el tiempo de ejecución sigue una ley potencial $T(N) = a \cdot N^b$, entonces:

$$\frac{T(2N)}{T(N)} = \frac{a\,(2N)^b}{a\,N^b} = 2^b$$

Por lo tanto, tomando logaritmo en base 2:

$$b = \log_2\!\left(\frac{T(2N)}{T(N)}\right)$$

Cuando la razón $R$ se **estabiliza** en un valor constante, podemos confiar en que hemos encontrado el orden de crecimiento real.

| Razón $R = T(2N)/T(N)$ estabilizada en... | Exponente $b$ | Complejidad |
|:---:|:---:|:---:|
| ~2 | 1 | $\mathcal{O}(n)$ |
| ~4 | 2 | $\mathcal{O}(n^2)$ |
| ~8 | 3 | $\mathcal{O}(n^3)$ |

> 💡 **Importante:** Para que este método funcione, los tamaños deben ser **estrictamente una secuencia de duplicación** (cada uno es el doble del anterior). Nuestros tamaños `[1000, 2000, 4000, 8000, 16000]` cumplen exactamente esta condición.

**Instrucciones:**  
Ejecuta la celda siguiente. Analiza la columna `log₂(razón)` y responde: ¿hacia qué valor converge? ¿Es consistente con la complejidad teórica $\mathcal{O}(n^2)$?

````python
# ============================================================
# Tabla de duplicación de Sedgewick
# ============================================================

print("Tabla de Hipótesis de Duplicación (Sedgewick)")
print("=" * 60)
print(f"{'N':>8}  {'T(N) (s)':>12}  {'T(2N)/T(N)':>12}  {'log₂(razón)':>13}")
print("-" * 60)

for i in range(len(tamanios)):
    n = tamanios[i]
    t = tiempos[i]
    
    if i == 0:
        # Para el primer tamaño no hay T(N/2) previo
        print(f"{n:>8}  {t:>12.6f}  {'---':>12}  {'---':>13}")
    else:
        t_prev = tiempos[i - 1]        # T(N/2)
        razon  = t / t_prev            # T(N) / T(N/2) = T(2·N_prev) / T(N_prev)
        log2_r = math.log2(razon)      # log₂(razón) ≈ b
        print(f"{n:>8}  {t:>12.6f}  {razon:>12.4f}  {log2_r:>13.4f}")

print("=" * 60)
print()
print("Interpretación:")
print("  · Si log₂(razón) ≈ 2  →  T(N) ∈ O(N²)  ✅ esperado para burbuja")
print("  · Si log₂(razón) ≈ 1  →  T(N) ∈ O(N)")
print("  · Si log₂(razón) ≈ 3  →  T(N) ∈ O(N³)")
````

## 📈 Parte 7: Análisis Log-Log y Regresión Lineal

### Relación entre la hipótesis de duplicación y el gráfico log-log

El método de duplicación de Sedgewick y el análisis log-log son **equivalentes matemáticamente**: ambos estiman el mismo exponente $b$. La diferencia es que el gráfico log-log permite visualizar todos los puntos a la vez y ajustar una recta por mínimos cuadrados, lo que es más robusto ante ruido experimental.

Si $T(n) = a \cdot n^b$, tomando $\log_{10}$ a ambos lados:

$$\log_{10} T(n) = \log_{10} a + b \cdot \log_{10} n$$

Esta es una recta con **pendiente $b$** en el espacio $(\log_{10} n,\; \log_{10} T)$.

> 📌 **Nota sobre la base:** La base del logaritmo **no afecta la pendiente**. Usar $\log_{10}$, $\log_2$ o $\ln$ produce exactamente la misma pendiente, porque el factor de cambio de base se cancela en el cociente.

**Completa el código en las celdas siguientes.**

````python
# ============================================================
# Calcular los logaritmos (base 10)
# ============================================================

# Calcula el logaritmo base 10 de cada tamaño
log_n = [math.log10(n) for n in tamanios]

# Calcula el logaritmo base 10 de cada tiempo medido
log_t = [math.log10(t) for t in tiempos]

print(f"{'log10(n)':>12}  {'log10(t)':>14}")
print("-" * 30)
for ln, lt in zip(log_n, log_t):
    print(f"{ln:>12.4f}  {lt:>14.4f}")
````

````python
# ============================================================
# Paso 7.2: Ajuste lineal (regresión) en escala log-log
# ============================================================

# numpy.polyfit ajusta un polinomio de grado 1 (línea recta)
# a los datos (log_n, log_t) por mínimos cuadrados.
# Retorna [pendiente, intercepto]

coeficientes = np.polyfit(log_n, log_t, 1)
pendiente    = coeficientes[0]
intercepto   = coeficientes[1]

print(f"Pendiente  (b) = {pendiente:.4f}")
print(f"Intercepto (a) = {10**intercepto:.6f}")
print(f"\nModelo ajustado: T(n) ≈ {10**intercepto:.2e} · n^{pendiente:.2f}")
````

````python
# ============================================================
# Paso 7.3: Gráfico log-log
# ============================================================

# Genera los puntos de la recta ajustada
log_n_array  = np.array(log_n)
recta_ajuste = pendiente * log_n_array + intercepto

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- Gráfico 1: Escala lineal ---
axes[0].plot(tamanios, tiempos, 'o-', color='steelblue',
             linewidth=2, markersize=8, label='Tiempo medido')
axes[0].set_xlabel('Tamaño del arreglo (n)', fontsize=12)
axes[0].set_ylabel('Tiempo de ejecución (s)', fontsize=12)
axes[0].set_title('Tiempo vs. Tamaño (escala lineal)', fontsize=13)
axes[0].legend()
axes[0].grid(True, linestyle='--', alpha=0.6)

# --- Gráfico 2: Escala log-log ---
axes[1].plot(log_n, log_t, 'o', color='steelblue',
             markersize=8, label='Datos observados')
axes[1].plot(log_n_array, recta_ajuste, '--', color='tomato',
             linewidth=2, label=f'Ajuste lineal (pendiente={pendiente:.2f})')
axes[1].set_xlabel('log₁₀(n)', fontsize=12)
axes[1].set_ylabel('log₁₀(T)', fontsize=12)
axes[1].set_title('Gráfico log-log: T vs. n', fontsize=13)
axes[1].legend()
axes[1].grid(True, linestyle='--', alpha=0.6)

plt.suptitle('Análisis de Complejidad — Ordenamiento por Burbuja', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_burbuja_loglog.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ Gráfico guardado como 'grafico_burbuja_loglog.png'")
````
